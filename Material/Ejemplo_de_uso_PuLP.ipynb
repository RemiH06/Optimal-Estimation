```python
# Importamos
import pulp as p

# Creamos un problema de Optimizacion
Lp_prob = p.LpProblem('Problem', p.LpMinimize)

# Create problem Variables
x = p.LpVariable("x", lowBound = 0, )   # Variable x >= 0
y = p.LpVariable("y", lowBound = 0)   # Variable y >= 0
```

```python
# Funcion objetivo
Lp_prob += 3 * x + 5 * y

# Restricciones
Lp_prob += 2 * x + 3 * y >= 12
Lp_prob += -x + y <= 3
Lp_prob += x >= 4
Lp_prob += y <= 3

# Imprimir el problema
print(Lp_prob)
```

```python
status = Lp_prob.solve()   # Ejecutar solucion
print(p.LpStatus[status])   # Status

# Imprimir la solucion
print(p.value(x), p.value(y), p.value(Lp_prob.objective))
```

```python
import pulp as lp

# Creamos problema (Como en cualquier modelo)

problema = lp.LpProblem(
    "Planeacion_de_produccion",
    lp.LpMaximize
)


# Definimos variables a optimizar

x_E = lp.LpVariable(
    "Material_Estandar",
    lowBound=0
)

x_P = lp.LpVariable(
    "Material_Premium",
    lowBound=0
)


# Definimos modelo o funcion objetivo
problema += (
    50*x_E + 40*x_P,
    "Margen_total"
)


# Determinamos restricciones

# Capacidad de laminación
problema += (
    2*x_E + x_P <= 100,
    "Laminacion"
)

# Capacidad de acabado
problema += (
    x_E + 2*x_P <= 80,
    "Acabado"
)


# Visualizamos antes de resolver
print(problema)
```

```python
# Resolver el problema
problema.solve()

# Mostrar el estado de la solución
print("Estado:", lp.LpStatus[problema.status])

# Mostrar los valores optimos
print("\nSolución óptima:")
print(f"Material estándar: {x_E.varValue:.2f} toneladas")
print(f"Material premium:  {x_P.varValue:.2f} toneladas")

# Mostrar el valor de la función objetivo
print(
    f"Margen máximo: ${lp.value(problema.objective):,.2f} "
    "miles de pesos"
)
```
